```python
# Normalizar respuestas
def normalizar_respuesta(texto):
    texto = texto.strip().lower()
    texto = unicodedata.normalize("NFD", texto)

    texto = "".join(
        letra for letra in texto
        if unicodedata.category(letra) != "Mn"
    )

    return texto


# Preparar preguntas sin repeticiones
def preparar_preguntas():
    preguntas_disponibles = {}

    for categoria, lista in preguntas.items():
        preguntas_disponibles[categoria] = lista.copy()
        random.shuffle(preguntas_disponibles[categoria])

    return preguntas_disponibles


# Hacer una pregunta
def hacer_pregunta(categoria, preguntas_disponibles):

    if not preguntas_disponibles[categoria]:
        preguntas_disponibles[categoria] = preguntas[categoria].copy()
        random.shuffle(preguntas_disponibles[categoria])

    pregunta_elegida = preguntas_disponibles[categoria].pop()

    print("\nCATEGORÍA:", categoria)

    respuesta_jugador = input(
        pregunta_elegida["pregunta"] + "\nTu respuesta: "
    )

    respuesta_correcta = pregunta_elegida["respuesta"]

    if normalizar_respuesta(respuesta_jugador) == normalizar_respuesta(respuesta_correcta):

        print("\n" + random.choice(mensajes_correctos))
        return True

    else:
        print("\n" + random.choice(mensajes_incorrectos))
        print("La respuesta correcta era:", respuesta_correcta)
        return False
```

```python
# Lanzar el dado
def lanzar_dado():
    dado = random.randint(1, 6)
    print("Has sacado un", dado)
    return dado

# Mover al jugador
def mover_jugador(posicion, dado):
    nueva_posicion = posicion + dado

    if nueva_posicion > FIN:
        print("¡Te has pasado de la casilla FINAL!")
        print("Necesitas el número exacto para ganar.")
        return posicion

    return nueva_posicion
```

```python
def jugar():

    # Bienvenida, reglas y tablero
    bienvenida()
    mostrar_reglas()
    mostrar_tablero()

    # Registrar jugadores
    jugador1, jugador2 = registrar_jugadores()
    jugadores = [jugador1, jugador2]

    # Posiciones iniciales
    posiciones = [0, 0]

    # Preparar preguntas
    preguntas_disponibles = preparar_preguntas()

    turno = 0

    print("\n¡COMIENZA BIOTRIVIAL UAX!")

    # Mostrar las fichas en INICIO
    mostrar_fichas(posiciones, jugadores)

    while True:

        jugador = jugadores[turno]

        print("\n================================")
        print("Turno de:", jugador)
        print("Posición actual:", posiciones[turno])

        # Si el jugador ya está en FINAL
        if posiciones[turno] == FIN:

            print("\n¡Ya estás en la casilla FINAL!")
            print("¡Tienes otra oportunidad para ganar!")

            categoria_final = random.choice(
                list(categorias.values())
            )

            print("Categoría final:", categoria_final)

            correcta = hacer_pregunta(
                categoria_final,
                preguntas_disponibles
            )

            if correcta:
                print("\n¡ENHORABUENA,", jugador, "!")
                print("¡HAS GANADO BIOTRIVIAL UAX!")
                print("¡Tus neuronas merecen unas vacaciones!")
                break

            else:
                print("¡Otra vez será!")
                turno = (turno + 1) % 2
                continue

        # Lanzar el dado
        input("\nPulsa ENTER para lanzar el dado...")

        dado = lanzar_dado()

        posicion_anterior = posiciones[turno]

        nueva_posicion = mover_jugador(
            posicion_anterior,
            dado
        )

        # Comprobar si ha llegado a FINAL
        if nueva_posicion == FIN:

            posiciones[turno] = FIN

            # Mover la ficha a FINAL
            mostrar_fichas(posiciones, jugadores)

            print("\n¡HAS LLEGADO A LA CASILLA FINAL!")
            print("Pero todavía no has ganado...")
            print("¡Tienes que superar la PREGUNTA FINAL!")

            categoria_final = random.choice(
                list(categorias.values())
            )

            print("Categoría final:", categoria_final)

            correcta = hacer_pregunta(
                categoria_final,
                preguntas_disponibles
            )

            if correcta:
                print("\n¡ENHORABUENA,", jugador, "!")
                print("¡HAS GANADO BIOTRIVIAL UAX!")
                print("¡Tus neuronas merecen unas vacaciones!")
                break

            else:
                print("\n¡Tan cerca y tan lejos!")
                print("Tendrás otra oportunidad en tu próximo turno.")
                turno = (turno + 1) % 2
                continue

        # Comprobar si se ha pasado de FINAL
        if nueva_posicion == posicion_anterior:

            print("¡Te has pasado de la casilla FINAL!")
            print("Pierdes el turno.")

            turno = (turno + 1) % 2
            continue

        # Actualizar posición
        posiciones[turno] = nueva_posicion

        print("Nueva posición:", nueva_posicion)

        # Mostrar el tablero con las fichas actualizadas
        mostrar_fichas(posiciones, jugadores)

        # Obtener categoría de la casilla
        color = tablero[nueva_posicion - 1]
        categoria = categorias[color]

        print("Color:", color)
        print("Categoría:", categoria)

        # Hacer pregunta
        correcta = hacer_pregunta(
            categoria,
            preguntas_disponibles
        )

        if correcta:
            print("¡Puedes volver a tirar!")

        else:
            print("Turno del siguiente jugador.")
            turno = (turno + 1) % 2

    print("\nFIN DE LA PARTIDA")
    print("¡Gracias por jugar a BioTrivial UAX!")
```
